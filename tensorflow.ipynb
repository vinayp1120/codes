```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("rifatalam3/coffee-revenue-dataset")

print("Path to dataset files:", path)
```

```python
import os
print("Files in dataset directory:", os.listdir(path))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data = pd.read_csv(path + "/Coffe_sales.csv")
```

```python
print(data.head())
print(data.info())
print(data.describe())
```

```python
print(data["coffee_name"].unique())
```

```python
#on a given day,how many cups of particular type of coffee were sold?
for i in data["coffee_name"].unique():
    print(f"Total amount of {i} sold: {data.loc[data['coffee_name'] == i, 'coffee_name'].count()}")
    print(f"Total revenue from {i} sold: {data.loc[data['coffee_name'] == i, 'money'].sum()}")

```

```python
print(data)
```

```python
for i in data["coffee_name"].unique():
    plt.bar(i,data.loc[data['coffee_name'] == i, 'money'].sum())
    plt.xlabel("Coffee Type")
    plt.ylabel("Total Revenue")
plt.title("Total Revenue from Each Coffee Type")
plt.grid()
plt.legend()
#increase the width and space between the names of the coffee types
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
print(data.loc[data["coffee_name"]=="Latte", "coffee_name"])
```

```python
for i in data["coffee_name"].unique():
    data["total_revenue"]= data["money"]*data.loc[data["coffee_name"]==i, "coffee_name"].count()
print(data.columns)
```

```python
print(data["total_revenue"].count()==data.shape[0])
```

```python
import matplotlib.pyplot as plt

# Group by coffee type and sum revenue
revenue_by_coffee = data.groupby("coffee_name")["total_revenue"].sum()
print(revenue_by_coffee)

# Plot pie chart
plt.pie(
    x=revenue_by_coffee.values,
    labels=revenue_by_coffee.index,
    autopct="%1.1f%%"
)
plt.title("Distribution of Total Revenue by Coffee Type")
plt.show()
```

```python
print(data["Month_name"].unique())
```

```python
data['month'] = pd.Categorical(data['Month_name'], 
                               categories=['Jan','Feb','Mar','Apr','May','Jun',
                                           'Jul','Aug','Sep','Oct','Nov','Dec'],
                               ordered=True)

data = data.sort_values('month')
revenue_per_month = data.groupby("month")["total_revenue"].sum()
revenue_per_month.sort_index(inplace=True)  # Ensure months are in chronological order
plt.bar(revenue_per_month.index, revenue_per_month.values,)
plt.xlabel("Month")
plt.ylabel("Total Revenue")
plt.title("Total Revenue per Month")
plt.grid()
plt.legend()
plt.show()
```

```python

plt.pie(
    x=revenue_per_month.values,
    labels=revenue_per_month.index,
    autopct="%1.1f%%"
)
plt.title("Distribution of Total Revenue by Month")
plt.show()
```
