```python
import tensorflow as tf
import sys
import platform

# Print TensorFlow version
print(f"TensorFlow version: {tf.__version__}")

# Print Python version (two ways)
print(f"Python version (sys): {sys.version}")
print(f"Python version (platform): {platform.python_version()}")
```

```python
a = tf.constant(2)
b = tf.constant(3)
print(f"Tensor a: {a}")
print(f"Tensor b: {b}")
c = a+b
print(type(c))
print(f"Tensor c (a + b): {c}")
```

```python
a = tf.cos(30.0)
print(f"Cosine of 2: {a}")
```

```python
b = tf.constant(25,dtype=tf.float32)
print(f"Tensor b: {b}")
```

```python
c= tf.constant([1, 2, 3, 4, 5])#1d tensor
print(f"Tensor c: {c}",f"\nShape of c: {c.shape}")
d = tf.constant([[1, 2], [3, 4], [5, 6]])#2d tensor
print(f"Tensor d: {d}",f"\nShape of d: {d.shape}")
```

```python
mat1 = tf.constant([[1, 2], [3, 4]])
mat2 = tf.constant([[5, 6], [7, 8]])
result = tf.matmul(mat1, mat2)
print(f"Result of matrix multiplication:\n{result}",f"\nShape of result: {result.shape}")
```

```python
r = tf.maximum([1, 4], [2, 3])
print(f"Element-wise maximum:\n{r}")
```

```python
r = tf.maximum([1, 4,2, 3],y=0)
print(f"Element-wise maximum:\n{r}")
```
